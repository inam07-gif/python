```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("student_clustering.csv")
```

```python
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=df['cgpa'],y=df['iq'])
plt.plot()
```

```python
from sklearn.cluster import KMeans
```

```python
wcss=[]
```

```python
for i in range(1,11):
  km=KMeans(n_clusters=i)
  km.fit_predict(df)
  wcss.append(km.inertia_)
```

```python
plt.plot(wcss, marker='o')
plt.xlabel("Number of clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

```python
x=df.iloc[:,:].values
```

```python
y_pred=KMeans(n_clusters=4).fit_predict(x)
```

```python
plt.scatter(x[y_pred==0,0],x[y_pred==0,1],color='blue')
plt.scatter(x[y_pred==1,0],x[y_pred==1,1],color='red')
plt.scatter(x[y_pred==2,0],x[y_pred==2,1],color='green')
plt.scatter(x[y_pred==3,0],x[y_pred==3,1],color='purple')
```
